Can an outer Meta-Evolve run improve the recipe used by an inner researcher while private tasks, budgets, and promotion remain outside candidate authority? The [standalone chapter](https://github.com/salzubi401/meta-evolve/blob/main/docs/learn/autoresearch.md) contains the complete explanation.

```python
from pathlib import Path
import runpy

root = next(
    path for path in (Path.cwd(), *Path.cwd().parents)
    if (path / "pyproject.toml").exists()
)
lab = runpy.run_path(root / "examples/research/autoresearch/main.py")
print("lab loaded:", "run_outer" in lab)
```

## Run the outer loop

The seed and two successors each receive the same four inner benchmark tasks and the same five-evaluation inner budget. Public task scores are evidence; private quality is the selection metric.

```python
result = lab["run_outer"]()
print(lab["report"](result))
```

## Read the evidence boundary

The proposer derives each edit from a public-report evidence record. Private task details never enter that evidence. Held-out scores are calculated only after `result.best()` has fixed the selected recipe, and `promotion=none` is explicit.

```python
[
    (str(trial.artifact.value), tuple(item.kind for item in trial.evidence))
    for trial in result.trials()
    if trial.artifact is not None
]
```
